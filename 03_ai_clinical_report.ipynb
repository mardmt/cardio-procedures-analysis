```python
import pandas as pd

df = pd.read_csv(r'F:\GITHUB\Cardio-Procedures-Analysis\heart_failure_clinical_records_dataset.csv')
print('Data loaded successfully')
```

```python
# Collecting findings from our analysis
total_patients = len(df)
avg_age = round(df['age'].mean(), 1)
mortality_rate = round((df['DEATH_EVENT'].sum() / total_patients) * 100, 1)
avg_ef = round(df['ejection_fraction'].mean(), 1)
avg_creatinine = round(df['serum_creatinine'].mean(), 1)

print(f'Total patients: {total_patients}')
print(f'Average age: {avg_age}')
print(f'Mortality rate: {mortality_rate}%')
print(f'Average ejection fraction: {avg_ef}%')
print(f'Average serum creatinine: {avg_creatinine}')
```

```python
def generate_clinical_report(total_patients, avg_age, mortality_rate, avg_ef, avg_creatinine):
    
    report = f"""
CLINICAL SUMMARY REPORT
=======================
Dataset: Heart Failure Clinical Records
Analysis Date: 2026

PATIENT POPULATION
------------------
Total Patients: {total_patients}
Average Age: {avg_age} years
Average Ejection Fraction: {avg_ef}%
Average Serum Creatinine: {avg_creatinine} mg/dL

MORTALITY ANALYSIS
------------------
Overall Mortality Rate: {mortality_rate}%
Survival Rate: {round(100 - mortality_rate, 1)}%
Clinical Note: A {mortality_rate}% mortality rate reflects the severe 
prognosis associated with heart failure in adults over 60.

KEY FINDINGS
------------
1. Peak risk window identified: 50-70 years of age
2. Ejection fraction below 38% associated with higher mortality
3. Serum creatinine identified as strongest predictor of death
4. Cardiorenal syndrome pattern observed in non-surviving patients

MODEL PERFORMANCE
-----------------
Logistic Regression Accuracy: 65% | Death Recall: 28%
Random Forest Accuracy: 72%      | Death Recall: 56%
Recommended Model: Random Forest
Limitation: 44% of deaths still missed - not suitable for clinical 
use without further development and validation

RECOMMENDATIONS
---------------
1. Prioritise monitoring of serum creatinine in heart failure patients
2. Expand dataset for improved model performance
3. Include all 13 clinical variables in future modelling
4. Validate model on external dataset before clinical application
    """
    return report
```

```python
report = generate_clinical_report(
    total_patients, 
    avg_age, 
    mortality_rate, 
    avg_ef, 
    avg_creatinine
)

print(report)
```
